```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#using a 2x8 df
d1 = [
    [4,5,6,7,2,-1,0,-1],
    [1,4,3,4,-1,-2,-3,-4]
]
df = pd.DataFrame(d1)      
```

```python
n = df.shape[1]
```

```python
plt.scatter(df.iloc[0], df.iloc[1])
plt.axhline(0, color = 'k')
plt.axvline(0, color = 'k')
df_mean = df.mean(axis =1)
plt.scatter(df_mean[0], df_mean[1], color = 'r')
```

```python
# Centering the data:-
def center(df):
    return df.sub(df_mean, axis=0)
df_centered = center(df)
df_centered
```

```python
plt.scatter(df_centered.iloc[0], df_centered.iloc[1])
plt.axhline(0, color = 'k')
plt.axvline(0, color = 'k')
df_centered_mean = df_centered.mean(axis =1)
plt.scatter(df_centered_mean[0], df_centered_mean[1], color = 'r')
```

```python
#Calculating covariance matrix
#formula = 1/n * sum(xi*x^t)
Cov_matrix = (df_centered @ df_centered.T) /n
Cov_matrix
```

```python
#Compute eigen value
eigenvalues, eigenvectors = np.linalg.eig(Cov_matrix.to_numpy())
```
